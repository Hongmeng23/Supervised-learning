```python
df.isnull().sum().sort_values(ascending=False)
```

```python
missing = df.isnull().sum().sort_values(ascending=False) 
missing_pct = (df.isnull().mean()*100).sort_values(ascending=False)

missing_report = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct.round(2)
})

missing_report
```

```python
for col in ["age", "annual_income", "monthly_spend"]:
    df[col] =df[col].fillna(df[col].median())

df["contract_type"] =df["contract_type"].fillna("Unknown")
df.isnull().sum()
```

```python
print("Number of duplicated rows:", df.duplicated().sum())
df[df.duplicated(keep=False)].sort_values("customer_id")
```

```python
df = df.drop_duplicates()
print("Duplicate rows after cleaning:",df.duplicated().sum())
```

```python
print(df["region"].value_counts(dropna=False))
```

```python
df["region"] = (
    df["region"]
    .astype("string")
    .str.strip()
    .str.title()
)

df["region"].value_counts(dropna=False)
```

```python
print("Invalid ages: ")
display(df[df["age"] < 0])

print("Invalid support-call counts: ")
display(df[df["support_calls"] < 0])
```

```python
df.loc[df["age"] < 0, "age"] = np.nan
df.loc[df["support_calls"] < 0, "support_calls"] = np.nan

df["age"] = df["age"].fillna(df["age"].median())
df["support_calls"] = df["support_calls"].fillna(df["support_calls"].median())

df[["age", "support_calls"]].describe()
```

```python
Q1 = df["annual_income"].quantile(0.25)
Q3 = df["annual_income"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["annual_income"] < lower) |
    (df["annual_income"] > upper)
]

print("Lower bound:", lower)
print("Upper bound:", upper)

display(outliers)
```

```python
print("Missing Values:")
display(df.isnull().sum())

print("Duplicate rows:", df.duplicated().sum())

print("Negative ages:", (df["age"] < 0).sum())
print("Negative support calls:", (df["support_calls"] < 0).sum())

print("\nRegions:")
display(df["region"].value_counts())
```

```python
df.to_csv("customer_churn_cleande.csv", index=False)
print("Saved:", "customer_churn_cleaned.csv")
print("Final shape: ", df.shape)
```
